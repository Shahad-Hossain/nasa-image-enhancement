```python
from PIL import Image
import math
```

```python
input1 = "heart.jpg"
try:
    img1 = Image.open(input1).convert('RGB') # Convert to RGB to ensure consistent 3-channel access
except FileNotFoundError:
    print(f"Error: The file '{input1}' was not found.")
    exit()

input2 = "photo.jpg"
try:
   img2 = Image.open(input2).convert('RGB') # Convert to RGB to ensure consistent 3-channel access
except FileNotFoundError:
   print(f"Error: The file '{input2}' was not found.")
```

```python
pixels1 = img1.load()

width1, height1 = img1.size

pixels2 = img2.load()

width2, height2 = img2.size
```

```python
# a + b
for i in range(min(width1, width2)):
  for j in range(min(height1, height2)):
    r, g, b = pixels1[i, j];

    newr = r + pixels2[i, j][0];
    newg = g + pixels2[i, j][1];
    newb = b + pixels2[i, j][2];

    pixels1[i, j] = (newr, newg, newb)

img1.save("output.jpg")
```

```python
# grayscale
for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j];

    greyVal = int(0.2989 * r + 0.5870 * g + 0.1140 * b)

    pixels1[i, j] = (greyVal, greyVal, greyVal)

img1.save("output.jpg")

#for i in range(width2):
 # for j in range(height2):
  #  r, g, b = pixels2[i, j];
#
 #   greyVal = int(0.2989 * r + 0.5870 * g + 0.1140 * b)

 #   pixels2[i, j] = (greyVal, greyVal, greyVal)
```

```python
# a - b
for i in range(min(width1, width2)):
  for j in range(min(height1, height2)):
    r, g, b = pixels1[i, j];

    newr = r - pixels2[i, j][0];
    newg = g - pixels2[i, j][1];
    newb = b - pixels2[i, j][2];

    pixels1[i, j] = (newr, newg, newb)

img1.save("output.jpg")
```

```python
# b - a
for i in range(min(width1, width2)):
  for j in range(min(height1, height2)):
    r, g, b = pixels2[i, j];

    newr = r - pixels1[i, j][0];
    newg = g - pixels1[i, j][1];
    newb = b - pixels1[i, j][2];

    pixels2[i, j] = (newr, newg, newb)

img2.save("output.jpg")
```

```python
# a * b
for i in range(min(width1, width2)):
  for j in range(min(height1, height2)):
    r, g, b = pixels1[i, j];

    newr = r * pixels2[i, j][0];
    newg = g * pixels2[i, j][1];
    newb = b * pixels2[i, j][2];

    pixels1[i, j] = (newr, newg, newb)

img1.save("output.jpg")
```

```python
# a / b (if pixel in b equals 0 then just set it to the pixel in a)
for i in range(min(width1, width2)):
  for j in range(min(height1, height2)):
    r, g, b = pixels1[i, j];

    newr = r if pixels2[i, j][0] == 0 else int(r / pixels2[i, j][0]);
    newg = g if pixels2[i, j][1] == 0 else int(g / pixels2[i, j][1]);
    newb = b if pixels2[i, j][2] == 0 else int(b / pixels2[i, j][2]);

    pixels1[i, j] = (newr, newg, newb)

img1.save("output.jpg")
```

```python
# b / a (if pixel in b equals 0 then just set it to the pixel in a)
for i in range(min(width1, width2)):
  for j in range(min(height1, height2)):
    r, g, b = pixels2[i, j];

    newr = r if pixels1[i, j][0] == 0 else int(r / pixels1[i, j][0]);
    newg = g if pixels1[i, j][1] == 0 else int(g / pixels1[i, j][1]);
    newb = b if pixels1[i, j][2] == 0 else int(b / pixels1[i, j][2]);

    pixels2[i, j] = (newr, newg, newb)

img2.save("output.jpg")
```

```python
# log base 2 (a) (divides each log value by the maximum log value out of all the pixels and multiplies it by 255)
maxr = float('-inf')
maxg = float('-inf')
maxb = float('-inf')

for i in range(width1):#          ---|
  for j in range(height1):#          |
    r, g, b = pixels1[i, j];#        |
    r = math.log(r + 1, 2)#          |
    g = math.log(g + 1, 2)#          |------ Search for maximum value after performing the log function for r, g, and b
    b = math.log(b + 1, 2)#          |
#                                    |
    maxr = max(maxr, r)#             |
    maxg = max(maxg, g)#             |
    maxb = max(maxb, b)#          ---|

for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j];
    newr = int((math.log(r + 1, 2) / maxr) * 255)#        ---|       Get r, g, b to a number between 0 and 1 by dividing
    newg = int((math.log(g + 1, 2) / maxg) * 255)#           |------ the log by the maximum value found for r, g, and b,
    newb = int((math.log(b + 1, 2) / maxb) * 255)#        ---|       and then multiply the number b y 255. This makes
    #                                                                the differences between the pixels more recognizable

    pixels1[i, j] = (newr, newg, newb)

img1.save("output.jpg")
```

```python
# exp(a) (divides the pixel by 50 before applying exp())
for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j];
    r /= 46#              ---|
    g /= 46#                 |--- Divide r, g, and b by a constant so that
    b /= 46#              ---|    exp() won't give a very large number

    newr = int(math.exp(r))#    ---|
    newg = int(math.exp(g))#       |--- Perform exp() on the updated values
    newb = int(math.exp(b))#    ---|

    pixels1[i, j] = (newr, newg, newb)

img1.save("output.jpg")
```

```python
# exp(a)
MAX_VALUE = 255

for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j];

    newr = int(MAX_VALUE if math.exp(r) > MAX_VALUE else math.exp(r))#    ---|
    newg = int(MAX_VALUE if math.exp(g) > MAX_VALUE else math.exp(g))#       |---
    newb = int(MAX_VALUE if math.exp(b) > MAX_VALUE else math.exp(b))#    ---|

    pixels1[i, j] = (newr, newg, newb)

img1.save("output.jpg")
```

```python
#sqrt(a) (divides the square root of the pixel by the largest square root pixel value from the entire image and multiplies it by 255)
maxr = float('-inf')
maxg = float('-inf')
maxb = float('-inf')

for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j];
    r = math.sqrt(r)
    g = math.sqrt(g)
    b = math.sqrt(b)

    maxr = max(maxr, r)
    maxg = max(maxg, g)
    maxb = max(maxb, b)

for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j];
    newr = int((math.sqrt(r) / maxr) * 255)
    newg = int((math.sqrt(g) / maxg) * 255)
    newb = int((math.sqrt(b) / maxb) * 255)

    pixels1[i, j] = (newr, newg, newb)

img1.save("output.jpg")
```

```python
# cos(a) (divides the pixel by 255 and multiplies that number with 2pi, adds 1 to the result of cos(), and multiples that number by 255)
for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j]

    newr = int((math.cos(r/255 * 2 * math.pi) + 1) * (255/2))
    newg = int((math.cos(g/255 * 2 * math.pi) + 1) * (255/2))
    newb = int((math.cos(b/255 * 2 * math.pi) + 1) * (255/2))

    pixels1[i, j] = (newr, newg, newb)
img1.save("output.jpg")
```

```python
# tan(a) (clamps the pixels to a value between pi/2 and -pi/2 to perform tan() without an error)
maxr = float('-inf')
minr = float('inf')
maxg = float('-inf')
ming = float('inf')
maxb = float('-inf')
minb = float('inf')

for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j]

    newr = math.tan(r/255 * (math.pi / 1.1) - (math.pi / 2.1))
    newg = math.tan(g/255 * (math.pi / 1.1) - (math.pi / 2.1))
    newb = math.tan(b/255 * (math.pi / 1.1) - (math.pi / 2.1))

    maxr = max(maxr, newr)
    minr = min(minr, newr)
    maxg = max(maxg, newg)
    ming = min(ming, newg)
    maxb = max(maxb, newb)
    minb = min(minb, newb)
print(minr)
for i in range(width1):
  for j in range(height1):
    r, g, b = pixels1[i, j]

    newr = int(((math.tan(r/255 * (math.pi / 1.1) - (math.pi / 2.1)) - minr) / (maxr - minr)) * 255)
    newg = int(((math.tan(g/255 * (math.pi / 1.1) - (math.pi / 2.1)) - ming) / (maxg - ming)) * 255)
    newb = int(((math.tan(b/255 * (math.pi / 1.1) - (math.pi / 2.1)) - minb) / (maxb - minb)) * 255)

    pixels1[i, j] = (newr, newg, newb)
img1.save("output.jpg")
```

```python
# ((a/255) + (b/255)) * 255
for i in range(width1):
  for j in range(height1):
    r1, g1, b1 = pixels1[i, j]
    r2, g2, b2 = pixels2[i, j]
    newr = int((r1 / 510 + r2 / 510) * 255)
    newg = int((b1 / 510 + b2 / 510) * 255)
    newb = int((g1 / 510 + g2 / 510) * 255)

    pixels1[i, j] = (newr, newg, newb)
img1.save("output.jpg")
```
